```python
from sklearn.linear_model import LogisticRegression

# The Data
# Features count of (free) and count of (!)
X = [
    [0, 0], [1, 0], [0, 1],   # Safe emails (Low counts)
    [5, 3], [10, 5], [3, 8],  # Spam emails (High counts)
    [1, 0], [0, 2],           # More Safe
    [6, 4], [8, 9]            # More Spam
]

# Target: 0 = Safe, 1 = Spam
y = [0, 0, 0, 1, 1, 1, 0, 0, 1, 1]

# Istantiate the LogReg
model = LogisticRegression()

# Fit the model
model.fit (X, y)

print("Spam Filter Trained! 🤖")
```

```python
# New Email = 4 "free" and 4 "!"
new_email = [[4, 5]]

prediction = model.predict(new_email)

#prediction = probablity (how confident is the model)
probability = model.predict_proba(new_email)

print(f"Prediction: {prediction[0]}     ----------------(0) Safe, (1) Spam")
print(f"Probability: {probability[0][1]*100:.2f} %")
#
```
