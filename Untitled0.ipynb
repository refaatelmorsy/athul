```python
import pandas as pd
from sklearn.linear_model import LinearRegression
#create the dataset like the Excel sheet
data = {
    'Size': [1400, 1600, 1700, 1875, 1100, 1550, 2350, 2450, 1425, 1700],
    'Bedrooms': [3, 3, 2, 4, 2, 3, 4, 5, 3, 2],
    'Age' : [20, 15, 5, 5, 30, 60, 10, 5, 30, 5],
    'Price' : [245, 312, 279, 308, 199, 219, 405, 324, 250, 280]
}
#turn it into a pandas dataframe (table)
df = pd.DataFrame(data)
print("Here is our data:")
print(df)
```

```python
# 1- separate features (x) and target (y)
# we drop 'price' to get just imputs
x = df[['Size', 'Bedrooms', 'Age']]
y = df['Price']
# 2- instantiate the model
model = LinearRegression()

# 3- fit the model (this runs the math instantly)
model.fit(x, y)

# 4-
print("Training comblete the model has learned the math.")
```

```python

print(f"Base Price (Bias): ${model.intercept_: .2f}k")
print("Features Weights:")
print(f" - Size:      {model.coef_[0]: .4f}")
print(f" - Bedrooms:  {model.coef_[1]: .2f}")
print(f" - Age:       {model.coef_[2]: .2f} ")
```

```python
#define a new house features
# [Size, Bedrooms, and Age]
new_house = [[2500, 4, 10]]

#ask the model to predict
predicted_price = model.predict(new_house)

#print the prediction
print(f"we should list this house for ${predicted_price[0]: .2f}k")
```

```python
# Create a DataFrame with the same column names we used for training
new_house_df = pd.DataFrame([[2500, 4, 10]], columns=['Size', 'Bedrooms', 'Age'])

# Now predict using the DataFrame
predicted_price = model.predict(new_house_df)

print(f"We should list this house for: ${predicted_price[0]:.2f}k")
```
