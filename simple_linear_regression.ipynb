```python
# 1. Initialize
w = 0
b = 0
learning_rate = 0.01  # Small steps to avoid overshooting
epochs = 10000  # How many times we want to "learn" (loop)
n = len(X) # Number of data points (5)

# 2. The Learning Loop
for i in range(epochs):
    # A. Make Prediction
    y_pred = w * X + b

    # B. Calculate Gradients (The Calculus Part)
    # Derivative with respect to w: (-2/n) * sum(x * error)
    d_w = (-2/n) * np.sum(X * (y - y_pred))

    # Derivative with respect to b: (-2/n) * sum(error)
    d_b = (-2/n) * np.sum(y - y_pred)

    # C. Update Weights (Gradient Descent Step)
    w = w - (learning_rate * d_w)
    b = b - (learning_rate * d_b)

    # Print progress every 100 steps
    if i % 1000 == 0:
        mse = np.mean((y - y_pred)**2)
        print(f"Epoch {i}: w = {w:.2f}, b = {b:.2f}, MSE = {mse:.2f}")

print("-----------------------")
print(f"Final w: {w:.2f}")
print(f"Final b: {b:.2f}")
```
